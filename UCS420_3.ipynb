```python
def categorize_performance(rating):
    if(rating>=4.5):
        return "Excellent"
    elif(rating>=4.0):
        return "Good"
    else:
        return "Average"

df["Performace_Category"]=df["Rating"].apply(categorize_performance)
print(df)
```
