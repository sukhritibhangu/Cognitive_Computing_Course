```python
import random as rd
rd.seed(1024170300)
L = [rd.randint(100, 900)for i in range(100)]
print(L)
```

```python
odd=[x for x in L if x%2 !=0]
print(odd)
print("count is",len(odd))
```

```python
even=[x for x in L if x%2 ==0]
print(even)
print("count is",len(even))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
```

```python
prime_numbers = [x for x in L if is_prime(x)]
print(prime_numbers)
print("Count", len(prime_numbers))
```

```python
most_frequent=max(set(L),key=L.count)
print(most_frequent)
print("No. of occurences",L.count(most_frequent))
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 3]
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("A =", A)
print("B =", B)
```

```python
union=A.union(B)
print(union)
```

```python
intersection=A.intersection(B)
print(intersection)
```

```python
A_minus_B=A.difference(B)
B_minus_A=B.difference(A)
print("A-B",A_minus_B)
print("B-A",B_minus_A)
```

```python
symmetric = A.symmetric_difference(B)
print("Symmetric difference =", symmetric)
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
print("Original A:", A)
X = int(input("Enter a value to remove from A: "))
A.discard(X)
print("Updated A:", A)
```

```python
#using pop ,return value
dict1 = my_dict.copy()
removed_value = dict1.pop("branch")
print("Using pop():", dict1)
print("Removed value:", removed_value)
# using del, simply delete
dict2 = my_dict.copy()
del dict2["branch"]
print("Using del:", dict2)
```
