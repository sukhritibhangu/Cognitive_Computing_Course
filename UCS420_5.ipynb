Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:
a) Addition of 2 in all the element  
b) Multiply 3 with all the elements 
c) Divide every element by 2

```python
import numpy as np
a=np.array([1,2,3,4,5])
print(a)
b=a+2
print(b)
b=a*3
print(b)
b=a//2
print(b)
```

```python
arr = np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])
x = np.array([1,2,3,4,5,1,2,1,1,1]) 
most_freq=np.bincount(x).argmax()
indices=np.where(x==most_freq)[0]
print("Most frequent value in x",most_freq)
print("Indices of frequent values in x",indices)
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
most_freq2=np.bincount(y).argmax()
indices2=np.where(y==most_freq2)[0]
print("Most frequent value in y",most_freq2)
print("Indices of frequent values in y",indices2)
```

```python
sukh=np.array([np.linspace(10,100,25)])
print("Dimension:",sukh.ndim)
print("Shape:",sukh.shape)
print("Total Elements:",sukh.size)
print("Data type:",sukh.dtype)
print("Total number of bytes",sukh.nbytes)
print("Transpose:",sukh.reshape(5,5).T)
```

```python
ucs420_sukh=np.array([[10,20,30,40],[50,60,70,80],[90,15,20,35]])
print("Mean:",np.mean(ucs420_sukh))
print("Median:",np.median(ucs420_sukh))
print("MAX:",np.max(ucs420_sukh))
print("MIN:",np.min(ucs420_sukh))
print("Unique elements:",np.unique(ucs420_sukh))
reshaped_ucs420_sukh = ucs420_sukh.reshape(4, 3)
print("Reshaped Array:")
print(reshaped_ucs420_sukh)
resized_ucs420_sukh = np.resize(ucs420_sukh, (2, 3))
print("Resized Array:")
print(resized_ucs420_sukh)
```
